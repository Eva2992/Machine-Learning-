Univariate Analysis

```python
# univaiate value ( categorical )

survived = df['Survived'].value_counts()
print(survived)

survived_label= df['Survived'].unique()
print(survived_label)

plt.pie( survived , labels = survived_label , autopct="%1.1f")
plt.show()
```

```python
sns.heatmap(df.select_dtypes(include='number').corr(), annot=True, cmap='coolwarm')
```

```python
df['Pclass'].value_counts()
```

```python
sns.countplot( data =df , x=df['Pclass'] , hue = df['Survived'])
df['Pclass'].value_counts()/len(df) *100
```
